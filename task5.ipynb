```python
#Python Lists, Tuples, Sets and Dictionaries
#1. Lists
#A list is an ordered, changeable collection that allows duplicate values

# Student marks
marks = [85, 72, 90, 65, 88]

print("Original list:", marks)

# Adding an element
marks.append(95)
print("After append:", marks)

# Removing an element
marks.remove(65)
print("After remove:", marks)

# Searching for an element
print("Is 90 present?", 90 in marks)

# Sorting the list
marks.sort()
print("Sorted marks:", marks)

# Updating an element
marks[0] = 75
print("After updating first element:", marks)
```

```python
#2. Tuples

#A tuple is an ordered, unchangeable collection that allows duplicate values.

# Employee information
employee = ("E101", "Rahul", "Data Analyst", 35000)

print("Employee details:", employee)

# Accessing elements
print("Employee ID:", employee[0])
print("Employee Name:", employee[1])

# Searching for an element
print("Is Data Analyst present?", "Data Analyst" in employee)

# Counting occurrences
print("Count of Rahul:", employee.count("Rahul"))

# Finding index
print("Index of Data Analyst:", employee.index("Data Analyst"))
```

```python
#3. Sets
#A set is an unordered collection of unique elements. It automatically removes duplicate values.

# Student course registrations
courses = {"Python", "SQL", "Excel", "Python", "Power BI"}

print("Original set:", courses)

# Adding an element
courses.add("Statistics")
print("After add:", courses)

# Removing an element
courses.remove("Excel")
print("After remove:", courses)

# Searching for an element
print("Is SQL present?", "SQL" in courses)

# Updating the set
courses.update(["Machine Learning", "Tableau"])
print("After update:", courses)

# Removing duplicates from a list
student_ids = [101, 102, 101, 103, 102, 104]
unique_ids = set(student_ids)

print("Unique student IDs:", unique_ids)
```

```python
#4. Dictionaries
#A dictionary stores data in key-value pairs. Keys are used to access values.

# Employee dataset
employee = {
    "id": 101,
    "name": "Priya",
    "department": "Data Science",
    "salary": 45000
}

print("Employee details:", employee)

# Accessing values using keys
print("Employee Name:", employee["name"])
print("Department:", employee["department"])

# Adding a new key-value pair
employee["experience"] = 2
print("After adding experience:", employee)

# Updating a value
employee["salary"] = 50000
print("After updating salary:", employee)

# Searching for a key
print("Is salary present?", "salary" in employee)

# Removing a key-value pair
employee.pop("experience")
print("After removing experience:", employee)
```

```python
#5. Practical Student Dataset

students = [
    {"id": 1, "name": "Amit", "marks": 85},
    {"id": 2, "name": "Sneha", "marks": 92},
    {"id": 3, "name": "Rahul", "marks": 78},
    {"id": 4, "name": "Priya", "marks": 92}
]

# Display student names
print("Student Names:")
for student in students:
    print(student["name"])

# Find students scoring above 80
print("\nStudents scoring above 80:")
for student in students:
    if student["marks"] > 80:
        print(student["name"], "-", student["marks"])

# Find unique marks using a set
unique_marks = {student["marks"] for student in students}
print("\nUnique marks:", unique_marks)

# Sort students by marks
sorted_students = sorted(students, key=lambda x: x["marks"], reverse=True)
print("\nStudents sorted by marks:")
for student in sorted_students:
    print(student["name"], "-", student["marks"])
```
